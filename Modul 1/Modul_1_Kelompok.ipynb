## Konfigurasi Kelompok 2

```python
kelompok = 2
bagian_ditutup = ['mata', 'mulut pada foto mahasiswa']
print('Kelompok:', kelompok)
print('Bagian yang ditutup:', ', '.join(bagian_ditutup))
```

## Fungsi Masking

```python
def rect_from_ratio(image, x1, y1, x2, y2):
    h, w = image.shape[:2]
    return (int(x1 * w), int(y1 * h)), (int(x2 * w), int(y2 * h))
def draw_rect(image, ratio_box, color, thickness=-1):
    pt1, pt2 = rect_from_ratio(image, *ratio_box)
    cv2.rectangle(image, pt1, pt2, color, thickness)
def draw_circle(image, center_ratio, radius_ratio, color, thickness=-1):
    h, w = image.shape[:2]
    center = (int(center_ratio[0] * w), int(center_ratio[1] * h))
    radius = int(radius_ratio * min(w, h))
    cv2.circle(image, center, radius, color, thickness)
def add_label(image, text, ratio_xy, color=(255, 255, 255)):
    h, w = image.shape[:2]
    org = (int(ratio_xy[0] * w), int(ratio_xy[1] * h))
    cv2.putText(image, text, org, cv2.FONT_HERSHEY_SIMPLEX, 0.7, color, 2, cv2.LINE_AA)
```

## Koordinat Mask Kelompok 2

```python
mask_kelompok_2 = [
    ((0.13, 0.29, 0.22, 0.35), (255, 0, 0), 'Mata'),
    ((0.15, 0.41, 0.24, 0.47), (0, 180, 255), 'Mulut'),
]
```

## Hasil Masking KTM

```python
ktm_masked = ktm_rgb.copy()
for ratio_box, color, label in mask_kelompok_2:
    draw_rect(ktm_masked, ratio_box, color, thickness=-1)
    add_label(ktm_masked, label, (ratio_box[0] + 0.01, ratio_box[1] + 0.05))
plt.figure(figsize=(11, 6))
plt.imshow(ktm_masked)
plt.title('Hasil masking KTM - Kelompok 2')
plt.axis('off')
plt.show()
```

## Simpan Hasil

```python
output_path = 'ktm_masked_kelompok_2.png'
cv2.imwrite(output_path, cv2.cvtColor(ktm_masked, cv2.COLOR_RGB2BGR))
print('Hasil disimpan sebagai:', output_path)
```
