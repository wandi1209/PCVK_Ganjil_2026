## Fungsi Bantu

```python
from pathlib import Path
def bgr_to_rgb(img_bgr):
    return cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
def show_rgb(img_rgb, title='', figsize=(7, 5)):
    plt.figure(figsize=figsize)
    plt.imshow(img_rgb)
    plt.title(title)
    plt.axis('off')
    plt.show()
def show_gray(img_gray, title='', cmap='gray', figsize=(6, 4)):
    plt.figure(figsize=figsize)
    plt.imshow(img_gray, cmap=cmap)
    plt.title(title)
    plt.axis('off')
    plt.show()
def titik(img, x_ratio, y_ratio):
    h, w = img.shape[:2]
    return int(x_ratio * w), int(y_ratio * h)
```

## D1 - Membaca Citra dari Google Drive

```python
pcvk_dir = Path('/content/drive/MyDrive/PCVK')
if pcvk_dir.exists():
    print('Folder PCVK ditemukan:', pcvk_dir)
    print('Isi folder PCVK:', [p.name for p in pcvk_dir.iterdir()])
else:
    raise FileNotFoundError('Folder /content/drive/MyDrive/PCVK belum ditemukan. Jalankan cell Koneksi Google Drive terlebih dahulu.')
ktm_bgr = cv2.imread(ktm_path)
if ktm_bgr is None:
    raise FileNotFoundError(f'File KTM tidak dapat dibaca. Cek nama dan path file: {ktm_path}')
foto_bgr = cv2.imread(foto_aktivitas_path)
if foto_bgr is None:
    raise FileNotFoundError(f'File foto aktivitas tidak dapat dibaca. Cek nama dan path file: {foto_aktivitas_path}')
ktm_rgb = bgr_to_rgb(ktm_bgr)
foto_rgb = bgr_to_rgb(foto_bgr)
print('File KTM:', ktm_path)
print('KTM shape:', ktm_bgr.shape)
print('File foto aktivitas:', foto_aktivitas_path)
print('Foto aktivitas shape:', foto_bgr.shape)
```

Kesimpulan: persentase piksel gelap menunjukkan seberapa besar area KTM yang memiliki intensitas rendah. Nilainya bergantung pada pencahayaan, warna latar, dan objek pada foto KTM.

## Studi Kasus 1 - Label Identitas Adaptif pada Foto Dokumentasi

```python
def beri_label_identitas(img_rgb, nama, keterangan):
    hasil = img_rgb.copy()
    h, w = hasil.shape[:2]
    bar_h = max(48, int(0.10 * h))
    font_scale = max(0.7, min(w, h) / 900)
    thickness = max(2, min(w, h) // 450)
    padding_x = max(16, int(0.025 * w))
    y1 = h - bar_h

    cv2.rectangle(hasil, (0, y1), (w, h), (0, 0, 0), -1)
    cv2.putText(hasil, nama, (padding_x, y1 + int(0.42 * bar_h)), cv2.FONT_HERSHEY_SIMPLEX, font_scale, (255, 255, 255), thickness, cv2.LINE_AA)
    cv2.putText(hasil, keterangan, (padding_x, y1 + int(0.78 * bar_h)), cv2.FONT_HERSHEY_SIMPLEX, font_scale * 0.75, (255, 255, 0), thickness, cv2.LINE_AA)
    return hasil

label_foto = beri_label_identitas(foto_rgb, nama_mahasiswa, 'Dokumentasi kegiatan')
label_ktm = beri_label_identitas(ktm_rgb, nama_mahasiswa, 'Identitas mahasiswa')

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].imshow(label_foto)
axes[0].set_title('Label adaptif pada foto potret')
axes[0].axis('off')
axes[1].imshow(label_ktm)
axes[1].set_title('Label adaptif pada citra berbeda ukuran')
axes[1].axis('off')
plt.show()
```

## Studi Kasus 2 - Thumbnail Persegi 1:1 tanpa Mengubah Rasio

```python
def buat_thumbnail_persegi(img_rgb, output_size=800, warna_kanvas=(245, 245, 245), nama_toko='TOKO'):
    h, w = img_rgb.shape[:2]
    scale = output_size / max(h, w)
    new_w = int(w * scale)
    new_h = int(h * scale)
    resized = cv2.resize(img_rgb, (new_w, new_h), interpolation=cv2.INTER_AREA)

    canvas = np.full((output_size, output_size, 3), warna_kanvas, dtype=np.uint8)
    x_offset = (output_size - new_w) // 2
    y_offset = (output_size - new_h) // 2
    canvas[y_offset:y_offset + new_h, x_offset:x_offset + new_w] = resized

    bar_h = max(50, output_size // 12)
    cv2.rectangle(canvas, (0, output_size - bar_h), (output_size, output_size), (20, 20, 20), -1)
    font_scale = output_size / 900
    thickness = max(2, output_size // 400)
    cv2.putText(canvas, nama_toko, (output_size // 30, output_size - bar_h // 3), cv2.FONT_HERSHEY_SIMPLEX, font_scale, (255, 255, 255), thickness, cv2.LINE_AA)
    return canvas

citra_potret_rgb = bgr_to_rgb(cv2.imread(citra_potret_path))
citra_lanskap_rgb = bgr_to_rgb(cv2.imread(citra_lanskap_path))
thumb_potret = buat_thumbnail_persegi(citra_potret_rgb, nama_toko=nama_toko)
thumb_lanskap = buat_thumbnail_persegi(citra_lanskap_rgb, nama_toko=nama_toko)

fig, axes = plt.subplots(1, 2, figsize=(10, 5))
axes[0].imshow(thumb_potret)
axes[0].set_title('Thumbnail dari citra potret')
axes[0].axis('off')
axes[1].imshow(thumb_lanskap)
axes[1].set_title('Thumbnail dari citra lanskap')
axes[1].axis('off')
plt.show()
```
